Step - 3 : Checking missing values, duplicates, and inconsistencies

```python
# Missing values
df.isnull().sum()

# Missing values
df.isnull().sum()

# Duplicates
df.duplicated().sum()

# Unique values in key columns
df['Marital_Status'].unique()
df['Education'].unique()
```

Step - 4: Data Cleaning

```python
# Fill missing Income values with median
df['Income'].fillna(df['Income'].median(), inplace=True)


# Remove duplicates
df.drop_duplicates(inplace=True)
```

Step - 5: Standardizing Categorical Values

```python
df['Marital_Status'] = df['Marital_Status'].replace({
    'Together': 'Married',
    'Absurd': 'Single',
    'YOLO': 'Single',
    'Divorced': 'Single',
    'Widow': 'Single'
})
```

Step - 6: Date Formatting

```python
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], dayfirst=True)
```

Step - 7: Feature Engineering

```python
# Age
df['Age'] = 2026 - df['Year_Birth']

# Total Spending
df['Total_Spending'] = df[['MntWines','MntFruits','MntMeatProducts',
                          'MntFishProducts','MntSweetProducts','MntGoldProds']].sum(axis=1)

# Total Children
df['Children'] = df['Kidhome'] + df['Teenhome']
```

Step - 7: Dropping Unnecessary Columns

```python
df.drop(['ID'], axis=1, inplace=True)
```

Step - 8: Save CLeaned Dataset

```python
df.to_csv(r"C:\Users\cheta\Desktop\Customer-Data-Wrangling\data\cleaned\cleaned_data.csv", index=False)
```
